```python
# PHASE
# - Gas Phase
# - Aquatic Solution
# - Non-aquatic Solution
# - Molecular Crystal

class Tags_Phase(BaseModel):  # Phase of the molecule
    gas_phase: bool = Field(description="Molecule/dataset described in the gas phase.")
    aquatic_solution: bool = Field(description="Molecule/dataset described in an aqueous (aquatic, water) solution.")
    non_aquatic_solution: bool = Field(description="Molecule/dataset described in a non-aqueous (not water) solution/solvent.")
    molecular_crystal: bool = Field(description="Molecule/dataset described in a molecular crystal.")
    explanation: str = Field(description="Explanation of the tags.")
    confidence: float = Field(description="Confidence in the decision.")
    
# PART OF CHEMISTRY
# - Astrochemistry
# - Organic
# - Inorganic
# - Electrochemistry
# - Physical Chemistry
# - Solubility
# - Chemical Physics
# - Quantum Chemistry
# - Diatoms

class Tags_PartOfChemistry(BaseModel):  # Part of the chemistry of the molecule
    astrochemistry: bool = Field(description="Dataset relates to astrochemistry.")
    organic: bool = Field(description="Dataset relates to organic chemistry.")
    inorganic: bool = Field(description="Dataset relates to inorganic chemistry.")
    electrochemistry: bool = Field(description="Dataset relates to electrochemistry.")
    physical_chemistry: bool = Field(description="Dataset relates to physical chemistry.")
    solubility: bool = Field(description="Dataset targets solubility-related properties.")
    chemical_physics: bool = Field(description="Dataset relates to chemical physics.")
    quantum_chemistry: bool = Field(description="Dataset relates to quantum chemistry.")
    diatoms: bool = Field(description="Dataset specifically relates to diatoms")
    explanation: str = Field(description="Explanation of the tags.")
    confidence: float = Field(description="Confidence in the decision.")

# MOLECULENET
# - Quantum Chemistry
# - Chemical Physics
# - Biochemistry
# - Physiology

class Tags_MoleculeNet(BaseModel):  # Class of dataset in MoleculeNet
    quantum_chemistry: bool = Field(description="Dataset category matches MoleculeNet quantum chemistry tasks.")
    chemical_physics: bool = Field(description="Dataset category matches MoleculeNet chemical physics tasks.")
    biochemistry: bool = Field(description="Dataset category matches MoleculeNet biochemistry tasks.")
    physiology: bool = Field(description="Dataset category matches MoleculeNet physiology tasks.")
    explanation: str = Field(description="Explanation of the tags.")
    confidence: float = Field(description="Confidence in the decision.")

# STRUCTURES
# - Coordination Chemistry
# - Ion Liquids
# - Peptides
# - Proteins
# - Solutions


class Tags_Structures(BaseModel):  # Structure of the molecule
    coordination_chemistry: bool = Field(description="Dataset targets/uses coordination chemistry structures.")
    ion_liquids: bool = Field(description="Dataset targets/uses ionic liquids / ion-liquid systems.")
    peptides: bool = Field(description="Dataset targets peptides / peptide sequences.")
    proteins: bool = Field(description="Dataset targets proteins / protein-related structures.")
    solutions: bool = Field(description="Dataset targets/uses generic solution/solvent mixture systems.")
    explanation: str = Field(description="Explanation of the tags.")
    confidence: float = Field(description="Confidence in the decision.")

# REGULATION
# - Drug Discovery
# - Pesticides
# - Pollutants
# - Water Quality
# - Toxicity

class Tags_Regulation(BaseModel):  # Usage of dataset in government policy, laws, and regulation
    drug_discovery: bool = Field(description="Dataset/regulation context mentions drug discovery policy/oversight.")
    pesticides: bool = Field(description="Dataset/regulation context mentions pesticides.")
    pollutants: bool = Field(description="Dataset/regulation context mentions pollutants.")
    water_quality: bool = Field(description="Dataset/regulation context mentions water quality.")
    toxicity: bool = Field(description="Dataset/regulation context mentions toxicity/safety compliance.")
    explanation: str = Field(description="Explanation of the tags.")
    confidence: float = Field(description="Confidence in the decision.")

# LIFE SCIENCES
# - Metabolomics
# - Lipidomics
# - Toxicology
# - Epigenetics
# - Public Health
# - Pharmaceuticals
# - Anti-aging

class Tags_LifeSciences(BaseModel):  # Life sciences of the molecule
    metabolomics: bool = Field(description="Dataset relates to metabolomics.")
    lipidomics: bool = Field(description="Dataset relates to lipidomics.")
    toxicology: bool = Field(description="Dataset relates to toxicology (life-sciences framing).")
    epigenetics: bool = Field(description="Dataset relates to epigenetics.")
    public_health: bool = Field(description="Dataset relates to public health.")
    pharmaceuticals: bool = Field(description="Dataset relates to pharmaceuticals.")
    anti_aging: bool = Field(description="Dataset relates to anti-aging / longevity.")
    explanation: str = Field(description="Explanation of the tags.")
    confidence: float = Field(description="Confidence in the decision.")

# DRUG DISCOVERY
# - Drugs
# - Antibiotics
# - Antidepressants
# - Psychoactive
# - Anti-aging
# - Toxicology

class Tags_DrugDiscovery(BaseModel):  # Drug discovery of the molecule
    drugs: bool = Field(description="Dataset relates to drug discovery broadly.")
    antibiotics: bool = Field(description="Dataset relates specifically to antibiotics.")
    antidepressants: bool = Field(description="Dataset relates specifically to antidepressants.")
    psychoactive: bool = Field(description="Dataset relates specifically to psychoactive substances.")
    anti_aging: bool = Field(description="Dataset relates to anti-aging as a drug-discovery target area.")
    toxicology: bool = Field(description="Dataset relates to toxicology as part of drug discovery.")
    explanation: str = Field(description="Explanation of the tags.")
    confidence: float = Field(description="Confidence in the decision.")


# ENVIRONMENTAL
# - Toxicology
# - Biodiversity
# - Water Quality
# - Water Treatment
# - Pesticides

class Tags_Environmental(BaseModel):  # Environmental of the molecule
    toxicology: bool = Field(description="Dataset relates to toxicology in an environmental context.")
    biodiversity: bool = Field(description="Dataset relates to biodiversity.")
    water_quality: bool = Field(description="Dataset relates to water quality.")
    water_treatment: bool = Field(description="Dataset relates to water treatment.")
    pesticides: bool = Field(description="Dataset relates to pesticides.")
    explanation: str = Field(description="Explanation of the tags.")
    confidence: float = Field(description="Confidence in the decision.")

# MATERIALS
# - Polymer Science
# - Biomaterials
# - Corrosion Science
# - Sensors
# - Docking & Binding

class Tags_Materials(BaseModel):  # Materials of the molecule
    polymer_science: bool = Field(description="Dataset relates to polymer science.")
    biomaterials: bool = Field(description="Dataset relates to biomaterials.")
    corrosion_science: bool = Field(description="Dataset relates to corrosion science.")
    sensors: bool = Field(description="Dataset relates to sensors.")
    docking_binding: bool = Field(description="Dataset relates to docking and/or binding (e.g., binding affinity).")
    explanation: str = Field(description="Explanation of the tags.")
    confidence: float = Field(description="Confidence in the decision.")

# INDUSTRY
# - Farming
# - Oil & Gas
# - Mining
# - Cosmetics
# - Pharmaceutical Industry
# - Food Industry

class Tags_Industry(BaseModel):  # Industry of the molecule
    farming: bool = Field(description="Dataset relates to farming/agriculture industry.")
    oil_gas: bool = Field(description="Dataset relates to oil & gas industry.")
    mining: bool = Field(description="Dataset relates to mining industry.")
    cosmetics: bool = Field(description="Dataset relates to cosmetics industry.")
    pharmaceutical_industry: bool = Field(description="Dataset relates to pharmaceutical industry.")
    food_industry: bool = Field(description="Dataset relates to food industry.")
    explanation: str = Field(description="Explanation of the tags.")
    confidence: float = Field(description="Confidence in the decision.")

# ANALYTICAL
# - Mass Spectrometry
# - Electrochemistry
# - Toxicity

class Tags_Analytical(BaseModel):  # Analytical of the molecule
    mass_spectrometry: bool = Field(description="Dataset involves mass spectrometry.")
    electrochemistry: bool = Field(description="Dataset involves electrochemistry measurements.")
    toxicity: bool = Field(description="Dataset involves toxicity and toxicology measurements/assessments.")
    explanation: str = Field(description="Explanation of the tags.")
    confidence: float = Field(description="Confidence in the decision.")


```

```python
from typing import Annotated


class DatasetTags(BaseModel):
    phase: Optional[Tags_Phase] = None
    part_of_chemistry: Optional[Tags_PartOfChemistry] = None
    molecule_net: Optional[Tags_MoleculeNet] = None
    structures: Optional[Tags_Structures] = None
    regulation: Optional[Tags_Regulation] = None
    life_sciences: Optional[Tags_LifeSciences] = None
    drug_discovery: Optional[Tags_DrugDiscovery] = None
    environmental: Optional[Tags_Environmental] = None
    materials: Optional[Tags_Materials] = None
    industry: Optional[Tags_Industry] = None
    analytical: Optional[Tags_Analytical] = None


def merge_dataset_tags(
    left: Optional[DatasetTags],
    right: Optional[DatasetTags],
) -> Optional[DatasetTags]:
    if left is None:
        return right
    if right is None:
        return left
    ld, rd = left.model_dump(), right.model_dump()
    merged: dict = {}
    for k in ld:
        lv, rv = ld[k], rd[k]
        merged[k] = rv if rv is not None else lv
    return DatasetTags(**merged)


class DatasetTagsDataState(BaseModel):
    workspace_path: str
    key: str
    description_files: List[str] = Field(default_factory=list)
    full_description: str = ""
    field_name: str = ""
    tags: Annotated[Optional[DatasetTags], merge_dataset_tags] = None

```

```python
# to Agent: do not modify.
def store_state(state: DatasetTagsDataState):
    key_folder = Path(state.workspace_path) / state.key
    VersionedYaml(key_folder).create("tags", state.model_dump())
```

```python
# LangGraph-based tag extraction:
# - one LLM request per tag group
# - parallel branches using `Send`

import asyncio
import random

from langgraph.types import Send


detect_dataset_tag_group_prompt = ChatPromptTemplate.from_messages([
    ("system", """
You are an expert dataset classifier in chemistry/cheminformatics.

Task:
- Read dataset description.
- Return JSON that matches ONLY the requested tag-group schema.

Rules:
- Derive tags from logic and evidence in the description.
- Return JSON only.
""".strip()),
    ("user", """
Tag group: {tag_group}

Fields (boolean JSON keys; meanings from the schema):
{field_descriptions}

Dataset description:
\"\"\"{dataset_description}\"\"\"
""".strip()),
])


GROUP_FIELD_TO_MODEL: dict[str, type[BaseModel]] = {
    "phase": Tags_Phase,
    "part_of_chemistry": Tags_PartOfChemistry,
    "molecule_net": Tags_MoleculeNet,
    "structures": Tags_Structures,
    "regulation": Tags_Regulation,
    "life_sciences": Tags_LifeSciences,
    "drug_discovery": Tags_DrugDiscovery,
    "environmental": Tags_Environmental,
    "materials": Tags_Materials,
    "industry": Tags_Industry,
    "analytical": Tags_Analytical,
}


def _pydantic_field_descriptions_for_prompt(model_cls: type[BaseModel]) -> str:
    lines: list[str] = []
    for name, finfo in model_cls.model_fields.items():
        desc = (finfo.description or "").strip()
        lines.append(f"- {name}: {desc}" if desc else f"- {name}")
    return "\n".join(lines)


detect_description_summary_prompt = ChatPromptTemplate.from_messages([
    ("system", """
You compress raw dataset readme or joined file text into plain prose for taxonomy/tag classifiers.

Rules:
- Output only meaning: topic, domain, what records represent, stated purpose.
- Use plain sentences; no markdown, lists, tables, headings, or code.
- Strip technical noise: column/schema keys, long identifier dumps (InChIKey/SMILES/CAS), URLs, DOI links, \"Fields:\" tables, file inventories.
- If the input is almost only identifiers with no narrative, say briefly that it looks like a chemical identifier list without a richer description.
- Reply with the cleaned prose only; no preamble or labels.
""".strip()),
    ("user", """
Raw text:
\"\"\"{raw_description}\"\"\"
""".strip()),
])


def _coerce_state(state) -> DatasetTagsDataState:
    """Send() passes a dict; other nodes may receive DatasetTagsDataState."""
    return state if isinstance(state, DatasetTagsDataState) else DatasetTagsDataState.model_validate(state)


async def _ainvoke_with_backoff(structured_llm, messages, max_retries: int = 6):
    for attempt in range(max_retries + 1):
        try:
            return await structured_llm.ainvoke(messages)
        except Exception as e:
            msg = str(e)
            is_rate_limit = ("429" in msg) or ("RateLimitError" in msg)
            if (not is_rate_limit) or (attempt == max_retries):
                raise
            sleep_s = min(2 ** attempt, 16) + random.random() * 0.25
            await asyncio.sleep(sleep_s)


def _message_content_text(msg) -> str:
    c = getattr(msg, "content", None)
    if isinstance(c, str):
        return c
    if isinstance(c, list):
        parts: list[str] = []
        for block in c:
            if isinstance(block, str):
                parts.append(block)
            elif isinstance(block, dict) and isinstance(block.get("text"), str):
                parts.append(block["text"])
        return "".join(parts)
    return "" if c is None else str(c)


async def summarize_description_for_tagging(state: DatasetTagsDataState) -> dict:
    state = _coerce_state(state)
    raw = (state.full_description or "").strip()
    if not raw:
        return {"full_description": ""}
    cap = 24000
    snippet = raw[:cap] + ("\n[TRUNCATED]" if len(raw) > cap else "")
    messages = detect_description_summary_prompt.format_messages(raw_description=snippet)
    out = await _ainvoke_with_backoff(LLM_BASIC, messages)
    return {"full_description": _message_content_text(out).strip()}


def load_description_files(state: DatasetTagsDataState) -> dict:
    state = _coerce_state(state)
    key_folder = Path(state.workspace_path) / state.key

    detect = VersionedYaml(key_folder).read("detect_files")
    if detect is None:
        return {
            "description_files": [],
            "full_description": "",
        }

    detect_files_state, _meta = detect
    description_files = list(detect_files_state.get("description_files", []) or [])

    parts: list[str] = []
    for filename in description_files:
        p = key_folder / filename
        if not p.exists():
            continue
        text = p.read_text(encoding="utf-8", errors="ignore")
        parts.append(f"### {filename}\n{text.strip()}\n\n\n")

    return {
        "description_files": description_files,
        "full_description": "\n".join(parts).strip(),
    }


def dispatch_tag_groups(state: DatasetTagsDataState):
    state = _coerce_state(state)
    base = state.model_dump(
        include={"workspace_path", "key", "description_files", "full_description"}
    )
    return [
        Send("classify_tag_group", {**base, "field_name": field_name})
        for field_name in GROUP_FIELD_TO_MODEL
    ]


async def classify_tag_group(state: DatasetTagsDataState) -> dict:
    state = _coerce_state(state)
    field_name = state.field_name
    model_cls = GROUP_FIELD_TO_MODEL[field_name]

    desc = state.full_description or ""
    if len(desc) > 12000:
        desc = desc[:12000] + "\n[TRUNCATED]"

    structured_llm = LLM_BASIC.with_structured_output(model_cls)
    messages = detect_dataset_tag_group_prompt.format_messages(
        tag_group=field_name,
        field_descriptions=_pydantic_field_descriptions_for_prompt(model_cls),
        dataset_description=desc,
    )

    result = await _ainvoke_with_backoff(structured_llm, messages)
    return {"tags": DatasetTags(**{field_name: result})}


reflection_and_fix_prompt = ChatPromptTemplate.from_messages([
    ("system", """
You are an expert dataset classifier reviewing tag assignments for chemistry/cheminformatics datasets.

Task:
- Compare the dataset description with the candidate tag assignment (all groups).
- Fix false positives, false negatives, and contradictions between groups.
- Preserve or improve each subgroup's explanation and confidence so they match the revised booleans.

Rules:
- Use only evidence from the description; do not invent facts, but use resoning.
- `Toxicity` and `toxicology` are synonyms.
- Derive gas, aquatic or solid phases from the context.
- If molecule inside body or living organism, it is aquatic solution.
- Output must include every tag subgroup from the schema (full DatasetTags), each subgroup complete.
""".strip()),
    ("user", """
Dataset description:
\"\"\"{dataset_description}\"\"\"

Candidate tags (full JSON from prior classifiers):
\"\"\"{candidate_tags}\"\"\"
""".strip()),
])


async def reflection_and_fix(state: DatasetTagsDataState) -> dict:
    state = _coerce_state(state)
    t = state.tags
    if t is None or any(getattr(t, name) is None for name in GROUP_FIELD_TO_MODEL):
        return {}
    desc = state.full_description or ""
    if len(desc) > 12000:
        desc = desc[:12000] + "\n[TRUNCATED]"
    candidate_tags = t.model_dump_json(indent=2)
    structured_llm = LLM_BASIC.with_structured_output(DatasetTags)
    messages = reflection_and_fix_prompt.format_messages(
        dataset_description=desc,
        candidate_tags=candidate_tags,
    )
    result = await _ainvoke_with_backoff(structured_llm, messages)
    fixed = result if isinstance(result, DatasetTags) else DatasetTags.model_validate(result)
    store_state(state.model_copy(update={"tags": fixed}))
    return {"tags": fixed}


def finalize_tags(state: DatasetTagsDataState) -> dict:
    state = _coerce_state(state)
    t = state.tags
    if t is None or any(getattr(t, name) is None for name in GROUP_FIELD_TO_MODEL):
        return {}
    return {}
```

```python
# Build graph once; use the next cell for a single key or the following cell for all keys.

graph = StateGraph(DatasetTagsDataState)

graph.add_node("load_description_files", load_description_files)
graph.add_node("summarize_description_for_tagging", summarize_description_for_tagging)
graph.add_node("classify_tag_group", classify_tag_group)
graph.add_node("finalize_tags", finalize_tags)
graph.add_node("reflection_and_fix", reflection_and_fix)

graph.add_edge(START, "load_description_files")
graph.add_edge("load_description_files", "summarize_description_for_tagging")
graph.add_conditional_edges("summarize_description_for_tagging", dispatch_tag_groups)
graph.add_edge("classify_tag_group", "finalize_tags")
graph.add_edge("finalize_tags", "reflection_and_fix")
graph.add_edge("reflection_and_fix", END)

tags_app = graph.compile()

run_config = {**LF_CONFIG, "max_concurrency": 4}
```

```python
# Run for one key (skip if you only use the all-keys cell below)
KEY = "ASPECT"  # e.g. "ALGALTOX"

raw = await tags_app.ainvoke(
    DatasetTagsDataState(workspace_path=WORKSPACE, key=KEY),
    config=run_config,
)
final_state = (
    raw if isinstance(raw, DatasetTagsDataState) else DatasetTagsDataState.model_validate(raw)
)
final_state
```

```python
# Run all workspace keys (fault-tolerant: one failure does not stop the rest).
# Requires: run the graph compile cell first so `tags_app` and `run_config` exist.

import traceback

all_keys = sorted(
    p.name
    for p in Path(WORKSPACE).iterdir()
    if p.is_dir() and not p.name.startswith(".")
)

batch_tag_results: list[dict] = []
for key in all_keys[:]:
    try:
        raw = await tags_app.ainvoke(
            DatasetTagsDataState(workspace_path=WORKSPACE, key=key),
            config=run_config,
        )
        final = (
            raw
            if isinstance(raw, DatasetTagsDataState)
            else DatasetTagsDataState.model_validate(raw)
        )
        batch_tag_results.append({"key": key, "ok": True, "state": final})
        print(f"OK  {key}")
    except Exception as e:
        batch_tag_results.append(
            {
                "key": key,
                "ok": False,
                "error": repr(e),
                "traceback": traceback.format_exc(),
            }
        )
        print(f"FAIL {key}: {e!r}")

failed_keys = [r["key"] for r in batch_tag_results if not r["ok"]]
print(f"\nDone: {len(batch_tag_results)} keys, {len(failed_keys)} failed")
failed_keys
```
